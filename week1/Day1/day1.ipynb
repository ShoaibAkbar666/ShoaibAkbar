```python
def price_category(price):   # .apply
    if price > 100:
        return 'Expensive'
    else:
        return 'Cheap'
```

```python
OrderItems['price_category'] = OrderItems['price'].apply(price_category) 
```

```python
OrderItems[['price', 'price_category']].head()
```

```python
#vectorize
import numpy as np
OrderItems['price_category'] = np.where(
    OrderItems['price'] > 100,
    'Expensive',
    'Cheap'
)
```

```python
print(" .apply : ")
%timeit OrderItems['price'].apply(lambda x: 'Expensive' if x > 100 else 'Cheap')
```

```python
print("vectorized : ")
%timeit np.where(OrderItems['price'] > 100,'Expensive','Cheap')
```

```python
tables = {
    'Customers': Customers,
    'Geolocation': geolocation,
    'Order Items': OrderItems,
    'Order Payments': Order_payments,
    'Order Reviews': order_reviews,
    'Orders': Orders,
    'Products': products,
    'Sellers': sellers,
    'Product Category Translation': product_category
}

for name, df in tables.items():
    print(f"{name}: {df.shape}")
```

```python
orphan_order_items = OrderItems[
    ~OrderItems['order_id'].isin(Orders['order_id'])
]

print("Orphan OrderItems:", len(orphan_order_items))
```

```python
orphan_payments = Order_payments[
    ~Order_payments['order_id'].isin(Orders['order_id'])
]

print("Orphan Payments:", len(orphan_payments))
```

```python
orphan_reviews = order_reviews[
    ~order_reviews['order_id'].isin(Orders['order_id'])
]

print("Orphan Reviews:", len(orphan_reviews))
```

```python
orphan_customers = Orders[
    ~Orders['customer_id'].isin(Customers['customer_id'])
]

print("Orphan customer IDs:", len(orphan_customers))
```

```python
orphan_products = OrderItems[
    ~OrderItems['product_id'].isin(products['product_id'])
]

print("Orphan product IDs:", len(orphan_products))
```

```python
orphan_sellers = OrderItems[
    ~OrderItems['seller_id'].isin(sellers['seller_id'])
]

print("Orphan seller IDs:", len(orphan_sellers))
```

```python
checks = {
    'OrderItems → Orders': (
        ~OrderItems['order_id'].isin(Orders['order_id'])
    ).sum(),

    'Payments → Orders': (
        ~Order_payments['order_id'].isin(Orders['order_id'])
    ).sum(),

    'Reviews → Orders': (
        ~order_reviews['order_id'].isin(Orders['order_id'])
    ).sum(),

    'Orders → Customers': (
        ~Orders['customer_id'].isin(Customers['customer_id'])
    ).sum(),

    'OrderItems → Products': (
        ~OrderItems['product_id'].isin(products['product_id'])
    ).sum(),

    'OrderItems → Sellers': (
        ~OrderItems['seller_id'].isin(sellers['seller_id'])
    ).sum()
}

for relationship, count in checks.items():
    print(f"{relationship}: {count} orphan keys")
```
