```python
from rapidfuzz import process, fuzz

cities = sellers['seller_city'].dropna().unique().tolist()

near_duplicates = []

for city in cities:
    matches = process.extract(
        city,
        cities,
        scorer=fuzz.ratio,
        limit=5
    )
    
    for match, score, _ in matches:
        if city != match and score >= 90:
            near_duplicates.append((city, match, score))

near_duplicates[:50]
```

```python
#Task 3: Outlier Detection
#On price and freight_value, compute IQR bounds (Q1 - 1.5*IQR, Q3 + 1.5*IQR) and separately z-scores (|z| > 3) — compare how many points each method flags, they won't fully agree.
#For each flagged outlier group, actually look at a sample of the rows — is it a data entry error (price = 0.01 for a real product) or a genuinely expensive item (legitimate luxury product)? This distinction decides whether you cap, remove, or keep.
#Write your decision per column with reasoning — "kept because manual inspection showed these are real premium products, not errors" is a valid and expected answer.
```

```python
from scipy.stats import zscore

for col in ['price', 'freight_value']:
    
    Q1 = OrderItems[col].quantile(0.25)
    Q3 = OrderItems[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    iqr_outliers = OrderItems[
        (OrderItems[col] < lower) | (OrderItems[col] > upper)
    ]

    z_outliers = OrderItems[
        abs(zscore(OrderItems[col])) > 3
    ]

    print(f"\n{col}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower bound: {lower:.2f}")
    print(f"Upper bound: {upper:.2f}")
    print(f"IQR outliers: {len(iqr_outliers)}")
    print(f"Z-score outliers: {len(z_outliers)}")
```

```python
import pandas as pd
from rapidfuzz import process, fuzz


# ============================================================
# 1. MISSING VALUE CLEANING
# ============================================================

def clean_missing_values(products, orders, order_reviews):
    """
    Handle missing values using column-specific decisions.

    Products:
    - product_category_name -> "Unknown"
    - product_name_lenght -> 0
    - product_description_lenght -> 0
    - product_photos_qty -> 0
    - physical dimensions -> median

    Orders:
    - Delivery/approval dates are intentionally kept as NaN
      because missingness represents incomplete order stages.

    Order reviews:
    - review_comment_title -> "No title"
    """

    products = products.copy()
    orders = orders.copy()
    order_reviews = order_reviews.copy()

    # Products
    products['product_category_name'] = (
        products['product_category_name'].fillna('Unknown')
    )

    products['product_name_lenght'] = (
        products['product_name_lenght'].fillna(0)
    )

    products['product_description_lenght'] = (
        products['product_description_lenght'].fillna(0)
    )

    products['product_photos_qty'] = (
        products['product_photos_qty'].fillna(0)
    )

    for col in [
        'product_weight_g',
        'product_length_cm',
        'product_height_cm',
        'product_width_cm'
    ]:
        products[col] = products[col].fillna(
            products[col].median()
        )

    # Orders
    # Missing dates are intentionally retained.
    # They can represent canceled/incomplete orders.

    # Order reviews
    order_reviews['review_comment_title'] = (
        order_reviews['review_comment_title']
        .fillna('No title')
    )

    return products, orders, order_reviews


# ============================================================
# 2. DUPLICATE CLEANING
# ============================================================

def clean_duplicates(tables):
    """
    Remove exact duplicate rows from all tables.

    Exact duplicates contain no additional information.
    """

    cleaned_tables = {}

    for name, df in tables.items():
        cleaned_tables[name] = (
            df.drop_duplicates()
              .reset_index(drop=True)
        )

    return cleaned_tables


# ============================================================
# 3. INCONSISTENT NAMING CLEANING
# ============================================================

def clean_naming(sellers, geolocation):
    """
    Standardize city and state names by converting them to
    lowercase and removing leading/trailing whitespace.

    RapidFuzz is used to identify possible near-duplicate
    city names for manual review.
    """

    sellers = sellers.copy()
    geolocation = geolocation.copy()

    # Standardize seller names
    sellers['seller_city'] = (
        sellers['seller_city']
        .str.lower()
        .str.strip()
    )

    sellers['seller_state'] = (
        sellers['seller_state']
        .str.lower()
        .str.strip()
    )

    # Standardize geolocation names
    geolocation['geolocation_city'] = (
        geolocation['geolocation_city']
        .str.lower()
        .str.strip()
    )

    geolocation['geolocation_state'] = (
        geolocation['geolocation_state']
        .str.lower()
        .str.strip()
    )

    return sellers, geolocation


# ============================================================
# 4. OUTLIER DETECTION
# ============================================================

def detect_outliers(order_items):
    """
    Detect price and freight_value outliers using both
    IQR and Z-score methods.

    Outliers are retained because manual inspection showed
    that extreme values can represent legitimate products
    and shipping costs.
    """

    order_items = order_items.copy()

    results = {}

    for col in ['price', 'freight_value']:

        # IQR
        Q1 = order_items[col].quantile(0.25)
        Q3 = order_items[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        iqr_mask = (
            (order_items[col] < lower_bound) |
            (order_items[col] > upper_bound)
        )

        # Z-score
        mean = order_items[col].mean()
        std = order_items[col].std()

        z_scores = (order_items[col] - mean) / std

        z_mask = z_scores.abs() > 3

        results[col] = {
            'Q1': Q1,
            'Q3': Q3,
            'IQR': IQR,
            'lower_bound': lower_bound,
            'upper_bound': upper_bound,
            'IQR_outliers': int(iqr_mask.sum()),
            'Z_score_outliers': int(z_mask.sum())
        }

    # No rows are removed/capped because the outliers
    # were determined to be legitimate observations.

    return order_items, results


# ============================================================
# 5. DATA QUALITY REPORT
# ============================================================

def generate_quality_report(tables, stage):
    """
    Generate a data quality report containing:
    - row count
    - total null count
    - null percentage
    - exact duplicate count
    """

    report = []

    for name, df in tables.items():

        total_cells = df.shape[0] * df.shape[1]

        null_count = int(df.isnull().sum().sum())

        null_percentage = (
            null_count / total_cells * 100
            if total_cells > 0 else 0
        )

        duplicate_count = int(df.duplicated().sum())

        report.append({
            'Stage': stage,
            'Table': name,
            'Rows': len(df),
            'Null Count': null_count,
            'Null %': round(null_percentage, 2),
            'Duplicate Rows': duplicate_count
        })

    return pd.DataFrame(report)


# ============================================================
# 6. DATA TYPE CLEANING
# ============================================================

def clean_dtypes(orders, order_reviews):
    """
    Convert date columns to datetime format.
    """

    orders = orders.copy()
    order_reviews = order_reviews.copy()

    order_date_columns = [
        'order_purchase_timestamp',
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date',
        'order_estimated_delivery_date'
    ]

    for col in order_date_columns:
        orders[col] = pd.to_datetime(
            orders[col],
            errors='coerce'
        )

    review_date_columns = [
        'review_creation_date',
        'review_answer_timestamp'
    ]

    for col in review_date_columns:
        order_reviews[col] = pd.to_datetime(
            order_reviews[col],
            errors='coerce'
        )

    return orders, order_reviews


# ============================================================
# MAIN
# ============================================================

if __name__ == "__main__":

    # Load data
    product_category = pd.read_csv(
        "product_category_name_translation.csv"
    )

    sellers = pd.read_csv("sellers.csv")
    products = pd.read_csv("products.csv")
    Orders = pd.read_csv("Orders.csv")
    order_reviews = pd.read_csv("Order_reviews.csv")
    Order_payments = pd.read_csv("order_payments.csv")
    geolocation = pd.read_csv("geolocation.csv")
    OrderItems = pd.read_csv("Order_Items.csv")
    Customers = pd.read_csv("Customers.csv")


    # --------------------------------------------------------
    # BEFORE CLEANING REPORT
    # --------------------------------------------------------

    tables_before = {
        'product_category': product_category,
        'sellers': sellers,
        'products': products,
        'Orders': Orders,
        'order_reviews': order_reviews,
        'Order_payments': Order_payments,
        'geolocation': geolocation,
        'OrderItems': OrderItems,
        'Customers': Customers
    }

    before_report = generate_quality_report(
        tables_before,
        "Before"
    )


    # --------------------------------------------------------
    # CLEANING
    # --------------------------------------------------------

    products, Orders, order_reviews = clean_missing_values(
        products,
        Orders,
        order_reviews
    )

    sellers, geolocation = clean_naming(
        sellers,
        geolocation
    )

    Orders, order_reviews = clean_dtypes(
        Orders,
        order_reviews
    )


    tables_clean = {
        'product_category': product_category,
        'sellers': sellers,
        'products': products,
        'Orders': Orders,
        'order_reviews': order_reviews,
        'Order_payments': Order_payments,
        'geolocation': geolocation,
        'OrderItems': OrderItems,
        'Customers': Customers
    }


    tables_clean = clean_duplicates(tables_clean)

    OrderItems, outlier_results = detect_outliers(
        tables_clean['OrderItems']
    )

    tables_clean['OrderItems'] = OrderItems


    # --------------------------------------------------------
    # AFTER CLEANING REPORT
    # --------------------------------------------------------

    after_report = generate_quality_report(
        tables_clean,
        "After"
    )


    # --------------------------------------------------------
    # BEFORE / AFTER REPORT
    # --------------------------------------------------------

    quality_report = before_report.merge(
        after_report,
        on='Table',
        suffixes=('_Before', '_After')
    )

    print("\n========== DATA QUALITY REPORT ==========\n")

    print(
        quality_report[
            [
                'Table',
                'Rows_Before',
                'Rows_After',
                'Null %_Before',
                'Null %_After',
                'Duplicate Rows_Before',
                'Duplicate Rows_After'
            ]
        ].to_string(index=False)
    )


    # --------------------------------------------------------
    # OUTLIER REPORT
    # --------------------------------------------------------

    print("\n========== OUTLIER REPORT ==========\n")

    for col, result in outlier_results.items():

        print(f"\n{col}")

        print(
            f"IQR outliers: "
            f"{result['IQR_outliers']}"
        )

        print(
            f"Z-score outliers: "
            f"{result['Z_score_outliers']}"
        )


    # --------------------------------------------------------
    # SAVE CLEANED DATA
    # --------------------------------------------------------

    for name, df in tables_clean.items():

        filename = f"cleaned_{name}.csv"

        df.to_csv(
            filename,
            index=False
        )

    quality_report.to_csv(
        "data_quality_report.csv",
        index=False
    )

    print("\nCleaning completed.")
```
